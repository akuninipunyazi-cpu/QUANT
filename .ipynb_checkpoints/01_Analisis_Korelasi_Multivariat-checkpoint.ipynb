### **Objektif Phase 1:**
1. **Analisis Korelasi Linier (Pearson) & Monotonik (Spearman):**
   - Menghitung korelasi seluruh variabel makroekonomi dan pasar terhadap `Target_Volatilitas_Y`.
   - Mengidentifikasi variabel dengan korelasi positif tertinggi (*volatility drivers / panic transmitters*).
   - Mengidentifikasi variabel dengan korelasi terendah / negatif (*stabilizer / leverage effect*).
2. **Analisis Dinamika Perubahan (Returns, Delta, & Spreads):**
   - Menguji korelasi first-difference / returns (1-day & 5-day) dan selisih suku bunga (*Interest Rate Spread*).
3. **Lagged Cross-Correlation Analysis ($t-1$ s/d $t-5$):**
   - Menguji apakah informasi historis $t-1$ hingga $t-5$ mempertahankan sinyal prediktif (*leading indicator*) terhadap volatilitas IHSG di masa depan tanpa kebocoran data (*anti-leakage*).
4. **Sintesis & Rekomendasi Fitur Turunan (Task 1 Blueprint):**
   - Menetapkan landasan empiris untuk perumusan minimal 3 fitur turunan baru ($X_1, X_2, X_3$).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# Konfigurasi visualisasi profesional
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
plt.rcParams['font.size'] = 10
plt.rcParams['axes.titlesize'] = 12
plt.rcParams['axes.labelsize'] = 11
plt.rcParams['figure.titlesize'] = 14
plt.rcParams['figure.dpi'] = 120

print("Library dan konfigurasi visualisasi berhasil dimuat.")
```

## 1. Ingestion Data & Pemisahan Sampel Berlabel
Dataset memuat 4.266 baris data harian dari tahun 2010 hingga 2026. Data tahun 2025–2026 merupakan out-of-sample test set (nilai target $Y$ berstatus NaN). Untuk analisis korelasi historis yang valid, kita menggunakan data berlabel (2010–2024).

```python
# Ingestion Dataset
df = pd.read_csv('DATASET.csv')
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values('Date').reset_index(drop=True)

# Pemisahan data berlabel (Train & Val) dan data uji (Test Set)
labeled_mask = df['Target_Volatilitas_Y'].notnull()
df_labeled = df[labeled_mask].copy()
df_test = df[~labeled_mask].copy()

print(f"Total Baris Keseluruhan : {len(df):,}")
print(f"Data Berlabel (2010-2024): {len(df_labeled):,} baris ({df_labeled['Date'].min().strftime('%Y-%m-%d')} s/d {df_labeled['Date'].max().strftime('%Y-%m-%d')})")
print(f"Data Uji Test (2025-2026): {len(df_test):,} baris ({df_test['Date'].min().strftime('%Y-%m-%d')} s/d {df_test['Date'].max().strftime('%Y-%m-%d')})")

# Rangkuman statistik deskriptif data berlabel
raw_features = ['OIL', 'GOLD', 'USDIDR', 'SP500', 'IHSG', 'VIX', 'CPI', 'BI_rate', 'US_rate', 'Target_Volatilitas_Y']
df_labeled[raw_features].describe().T[['mean', 'std', 'min', '50%', 'max']]
```

## 2. Analisis Korelasi Linier (Pearson) & Monotonik (Spearman) pada Level Data
- **Pearson Correlation:** Mengukur hubungan linier langsung antara variabel independen dan target.
- **Spearman Rank Correlation:** Mengukur hubungan monotonik (mampu menangkap pola hubungan non-linear dan lebih tahan terhadap outlier ekstrem).

```python
# Matriks korelasi penuh
pearson_mat = df_labeled[raw_features].corr(method='pearson')
spearman_mat = df_labeled[raw_features].corr(method='spearman')

# Korelasi terhadap Target Volatilitas
corr_target = pd.DataFrame({
    'Pearson (r)': pearson_mat['Target_Volatilitas_Y'].drop('Target_Volatilitas_Y'),
    'Spearman (rho)': spearman_mat['Target_Volatilitas_Y'].drop('Target_Volatilitas_Y')
})
corr_target['Abs_Pearson'] = corr_target['Pearson (r)'].abs()
corr_target = corr_target.sort_values('Abs_Pearson', ascending=False).drop(columns=['Abs_Pearson'])

print("=== KORELASI LEVEL DATA TERHADAP Target_Volatilitas_Y ===")
display(corr_target.round(4))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Subplot 1: Heatmap Korelasi Pearson Penuh
mask = np.triu(np.ones_like(pearson_mat, dtype=bool))
sns.heatmap(pearson_mat, annot=True, fmt=".2f", cmap='vlag', center=0, 
            mask=mask, ax=axes[0], cbar_kws={'label': 'Pearson Correlation (r)'}, square=True)
axes[0].set_title('Matriks Korelasi Multivariat Antar Seluruh Variabel (Pearson)', pad=12, fontweight='bold')

# Subplot 2: Bar Komparasi Pearson vs Spearman terhadap Target
corr_target_plot = corr_target.sort_values('Pearson (r)', ascending=True)
y_pos = np.arange(len(corr_target_plot))
height = 0.35

axes[1].barh(y_pos - height/2, corr_target_plot['Pearson (r)'], height, label='Pearson (Linier)', color='#2b5c8f')
axes[1].barh(y_pos + height/2, corr_target_plot['Spearman (rho)'], height, label='Spearman (Monotonik)', color='#e26d5c')
axes[1].axvline(0, color='black', linestyle='--', linewidth=0.8, alpha=0.7)
axes[1].set_yticks(y_pos)
axes[1].set_yticklabels(corr_target_plot.index, fontweight='semibold')
axes[1].set_xlabel('Koefisien Korelasi terhadap Target Volatilitas IHSG')
axes[1].set_title('Korelasi Variabel Makro vs Target_Volatilitas_Y', pad=12, fontweight='bold')
axes[1].legend(loc='lower right', frameon=True)

plt.tight_layout()
plt.show()
```

### **Temuan Utama Analisis Level Data:**
1. **Pendorong Volatilitas Tertinggi (Positive Driver):**
   - **`VIX` (CBOE Volatility Index):** Menunjukkan korelasi positif paling dominan ($r = +0.4342$, $\rho = +0.2620$). Ini membuktikan secara kuantitatif adanya transmisi risiko global (*volatility contagion*): ketika bursa AS mengalami guncangan ketakutan, volatilitas pasar modal Indonesia melonjak serempak.
   - **`BI_rate`:** Menunjukkan korelasi positif ($r = +0.0782$, $\rho = +0.1057$). Saat suku bunga acuan BI tinggi (kebijakan moneter restriktif), pasar saham mengalami biaya modal lebih ketat dan rentan terhadap fluktuasi.
2. **Korelasi Negatif Terkuat (Stabilizer & Leverage Effect):**
   - **`IHSG` Level:** Menunjukkan korelasi negatif yang kuat ($r = -0.2842$, $\rho = -0.2694$). Ini mengonfirmasi fenomena pasar klasik: saat IHSG tertekan/turun (*bear market/correction*), kepanikan meningkat dan volatilitas membubung tinggi (*leverage effect*). Sebaliknya saat IHSG mencetak rekor tertinggi (bull market stabil), volatilitas cenderung terkompresi.
   - **`US_rate`:** Berkorelasi negatif ($r = -0.2005$, $\rho = -0.2410$). Suku bunga US yang rendah secara historis bertepatan dengan era pelonggaran kuantitatif (QE) paska krisis atau periode pandemi yang diiringi volatilitas pasar yang sangat tinggi.

## 3. Analisis Dinamika Finansial: Returns, Volatilitas Dinamik, & Interest Rate Spread
Dalam ekonometrika keuangan, harga nominal sering kali non-stasioner. Perubahan harga (returns 1-hari dan 5-hari) serta perbedaan suku bunga antarnegara (*interest rate differential*) sering kali menyimpan sinyal risiko yang jauh lebih sensitif.

```python
# Formulasi Fitur Dinamika Pasar Finansial
df_dyn = pd.DataFrame({'Date': df_labeled['Date'], 'Target_Volatilitas_Y': df_labeled['Target_Volatilitas_Y']})

# Return 1-hari (daily) dan 5-hari (weekly)
for col in ['OIL', 'GOLD', 'USDIDR', 'SP500', 'IHSG', 'VIX']:
    df_dyn[f'{col}_ret_1d'] = df_labeled[col].pct_change(1)
    df_dyn[f'{col}_ret_5d'] = df_labeled[col].pct_change(5)

# Interest Rate Spread (BI_rate - US_rate) dan perubahan bunga
df_dyn['rate_spread'] = df_labeled['BI_rate'] - df_labeled['US_rate']
df_dyn['delta_BI_rate'] = df_labeled['BI_rate'].diff(1)
df_dyn['delta_US_rate'] = df_labeled['US_rate'].diff(1)

# FX Pressure x VIX Interaction (Twin Shock Proxy)
df_dyn['fx_vix_shock'] = df_dyn['USDIDR_ret_5d'] * df_labeled['VIX']

# Hitung korelasi terhadap target
dyn_features = [c for c in df_dyn.columns if c not in ['Date', 'Target_Volatilitas_Y']]
dyn_corr = pd.DataFrame({
    'Pearson (r)': df_dyn[dyn_features].apply(lambda s: s.corr(df_dyn['Target_Volatilitas_Y'], method='pearson')),
    'Spearman (rho)': df_dyn[dyn_features].apply(lambda s: s.corr(df_dyn['Target_Volatilitas_Y'], method='spearman'))
})
dyn_corr['Abs_Pearson'] = dyn_corr['Pearson (r)'].abs()
dyn_corr = dyn_corr.sort_values('Abs_Pearson', ascending=False).drop(columns=['Abs_Pearson'])

print("=== KORELASI FITUR DINAMIKA/RETURNS TERHADAP TARGET VOLATILITAS ===")
display(dyn_corr.head(10).round(4))
```

```python
plt.figure(figsize=(12, 7))

# Visualisasi Top 12 Fitur Dinamika Finansial
top_dyn = dyn_corr.head(12).sort_values('Pearson (r)', ascending=True)
y_pos = np.arange(len(top_dyn))
bar_colors = ['#c0392b' if val < 0 else '#27ae60' for val in top_dyn['Pearson (r)']]

bars = plt.barh(y_pos, top_dyn['Pearson (r)'], color=bar_colors, alpha=0.85, edgecolor='black', linewidth=0.5)
plt.axvline(0, color='black', linestyle='--', linewidth=0.8)
plt.yticks(y_pos, top_dyn.index, fontweight='semibold')
plt.xlabel('Koefisien Korelasi Pearson (r) terhadap Target Volatilitas IHSG')
plt.title('Top Indikator Dinamika Finansial & Shock Drivers terhadap Volatilitas IHSG', pad=14, fontweight='bold')

# Anotasi nilai pada bar
for bar in bars:
    width = bar.get_width()
    offset = 0.005 if width >= 0 else -0.025
    plt.text(width + offset, bar.get_y() + bar.get_height()/4, f"{width:.3f}", 
             va='center', fontsize=9, fontweight='bold', color='black')

plt.xlim(-0.28, 0.28)
plt.tight_layout()
plt.show()
```

### **Temuan Analisis Dinamika Finansial:**
1. **`IHSG_ret_5d` ($r = -0.2295$): Asymmetric Volatility Shock**
   - Hubungan berbanding terbalik yang sangat signifikan: penurunan kumulatif IHSG selama 5 hari terakhir merupakan sinyal paling kuat terjadinya peningkatan volatilitas 5 hari ke depan. Penurunan tajam memicu *stop-loss liquidation* dan ketakutan institusional.
2. **`rate_spread` ($r = +0.2042$, $\rho = +0.2085$): Dominasi Spread Kebijakan Moneter**
   - Selisih antara suku bunga Bank Indonesia dan Federal Funds Rate (`BI_rate - US_rate`) berkorelasi positif kuat dan konsisten. Ini sejalan dengan premis stabilitas eksternal dan risiko *capital flight*.
3. **`VIX_ret_5d` ($r = +0.1524$): Akselerasi Kepanikan Global**
   - Bukan hanya level VIX, tetapi lonjakan VIX dalam rentang 5 hari memberikan sinyal akselerasi risiko yang menjalar ke bursa domestik.
4. **`SP500_ret_5d` ($r = -0.1698$): Transmisi Penurunan Ekuitas Global**
   - Pelemahan bursa Wall Street selama sepekan mentransmisikan sentimen *risk-off* yang memperlebar rentang fluktuasi IHSG.

## 4. Lagged Cross-Correlation Analysis ($t-1$ s/d $t-5$)
Dalam pemodelan *Early Warning System* anti-kebocoran data (*anti-leakage*), model pada hari $t$ hanya diizinkan melihat informasi masa lalu ($t, t-1, \dots, t-5$).  
Pertanyaan kuantitatif krusial: **Apakah korelasi variabel prediktor tetap bertahan kuat saat dimundurkan ke masa lalu ($t-1$ hingga $t-5$)?**

```python
# Analisis Lagged Cross-Correlation dari t-1 hingga t-5
lags = [1, 2, 3, 4, 5]
selected_vars = {
    'VIX': df_labeled['VIX'],
    'rate_spread': df_dyn['rate_spread'],
    'IHSG': df_labeled['IHSG'],
    'US_rate': df_labeled['US_rate'],
    'BI_rate': df_labeled['BI_rate'],
    'SP500': df_labeled['SP500'],
    'USDIDR': df_labeled['USDIDR'],
    'OIL': df_labeled['OIL'],
    'GOLD': df_labeled['GOLD'],
    'IHSG_ret_5d': df_dyn['IHSG_ret_5d'],
    'VIX_ret_5d': df_dyn['VIX_ret_5d']
}

lag_pearson_dict = {}
lag_spearman_dict = {}

target_series = df_labeled['Target_Volatilitas_Y']

for var_name, series in selected_vars.items():
    p_corrs = []
    s_corrs = []
    for lag in lags:
        lagged_s = series.shift(lag)
        valid = target_series.notnull() & lagged_s.notnull()
        p_corrs.append(stats.pearsonr(lagged_s[valid], target_series[valid])[0])
        s_corrs.append(stats.spearmanr(lagged_s[valid], target_series[valid])[0])
    lag_pearson_dict[var_name] = p_corrs
    lag_spearman_dict[var_name] = s_corrs

lag_pearson_df = pd.DataFrame(lag_pearson_dict, index=[f't-{i}' for i in lags]).T
lag_spearman_df = pd.DataFrame(lag_spearman_dict, index=[f't-{i}' for i in lags]).T

print("=== LAGGED PEARSON CORRELATION (t-1 s/d t-5) DENGAN TARGET Y ===")
display(lag_pearson_df.round(4))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Subplot 1: Heatmap Lagged Pearson
sns.heatmap(lag_pearson_df, annot=True, fmt=".3f", cmap='coolwarm', center=0, 
            ax=axes[0], cbar_kws={'label': 'Pearson Correlation (r)'})
axes[0].set_title('Heatmap Daya Prediktif Lag (t-1 s/d t-5) vs Target Volatilitas', pad=12, fontweight='bold')
axes[0].set_xlabel('Lag Hari Historis (t-k)')

# Subplot 2: Trajectory Line Plot untuk Top Leading Indicators
key_indicators = ['VIX', 'rate_spread', 'IHSG_ret_5d', 'VIX_ret_5d', 'SP500']
colors = ['#d9534f', '#0275d8', '#5cb85c', '#f0ad4e', '#6f42c1']

for ind, col in zip(key_indicators, colors):
    axes[1].plot(lags, lag_pearson_df.loc[ind, :], marker='o', linewidth=2.5, label=ind, color=col)

axes[1].axhline(0, color='black', linestyle='--', linewidth=0.8, alpha=0.7)
axes[1].set_xticks(lags)
axes[1].set_xticklabels([f't-{i}' for i in lags], fontweight='semibold')
axes[1].set_xlabel('Lag Waktu Prediktor')
axes[1].set_ylabel('Korelasi Pearson (r)')
axes[1].set_title('Daya Tahan Sinyal Prediksi (Signal Persistence) Sepanjang Lag', pad=12, fontweight='bold')
axes[1].legend(loc='center right', frameon=True)

plt.tight_layout()
plt.show()
```

### **Temuan Krusial Analisis Lagged Cross-Correlation:**
1. **Ketahanan Sinyal VIX (*High Persistence*):**
   - Korelasi Pearson VIX tetap berada di kisaran $+0.407$ pada $t-1$ dan $+0.360$ pada $t-5$. Ini menunjukkan bahwa lonjakan ketakutan di pasar keuangan global tidak reda dalam hitungan jam, melainkan memiliki efek rambatan berkepanjangan (*memory effect*).
2. **Kestabilan Sinyal Moneter (*Rate Spread Persistence*):**
   - `rate_spread` berada sangat stabil pada rentang $+0.204$ hingga $+0.207$ di seluruh lag $t-1$ sampai $t-5$. Variabel ini memberikan *structural risk baseline* yang kokoh.
3. **Penyusutan Bertahap Momentum (*Decaying Momentum Signal*):**
   - Fitur perubahan seperti `IHSG_ret_5d` dan `VIX_ret_5d` mengalami peluruhan korelasi bertahap dari $t-1$ ke $t-5$, yang mengonfirmasi sifat alaminya sebagai *short-term shock indicators*.

Berdasarkan analisis korelasi multivariat, sifat monotonik, dan ketahanan lag, berikut adalah rekomendasi **3 Fitur Turunan Baru ($X_1, X_2, X_3$)** untuk Task 1:

| Kode Fitur | Nama & Formulasi Fitur | Rasionalitas Bisnis & Finansial | Bukti Kuantitatif Fase 1 |

Spread sempit memicu tekanan likuiditas domestik. | Korelasi Pearson konsisten $+0.204$ dan Spearman $+0.208$, bertahan stabil sepanjang lag $t-1$ s/d $t-5$. |
