**Peneliti / Analis:** Senior Quantitative Risk & Machine Learning Analyst  
**Dataset:** `DATASET.csv` (Indonesia Financial Time Series 2010–2026)  
**Tujuan Analisis:**
1. Mengidentifikasi rezim kondisi makroekonomi dan sentimen pasar global secara objektif (*Unsupervised Market Regime Detection*).
2. Mengelompokkan data historis ke dalam klaster-klaster kondisi event ekonomi (misal: *Global Panic & Crisis*, *Aggressive Fed Tightening*, *High Spread / Low US Rate*, dan *Normal Transition*).
3. Membuktikan secara empiris apakah tingkat realized volatility IHSG (`Target_Volatilitas_Y`) terdistribusi berbeda secara signifikan di setiap rezim pasar.
4. Menghasilkan visualisasi institusional (*PCA 2D Cluster Space*, *Distribution Boxplots*, dan *Historical Regime Timeline Shading*).

---

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# Pastikan folder penyimpanan gambar lokal tersedia
os.makedirs('figures', exist_ok=True)

# Konfigurasi visualisasi
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
plt.rcParams['font.size'] = 10
plt.rcParams['axes.titlesize'] = 12
plt.rcParams['axes.labelsize'] = 11
plt.rcParams['figure.titlesize'] = 14
plt.rcParams['figure.dpi'] = 120

print("Library berhasil dimuat dan folder 'figures/' siap digunakan.")
```

## 1. Ingestion Data & Sanitasi Input Makro
Data berlabel (2010–2024) dimuat dan disortir kronologis. 
Pada tahap inspeksi kualitas data, ditemukan 2 entri desimal typo pada kurs USDIDR di tahun 2010 dan 2012 (nilai tercatat ~888 dan ~892 alih-alih ~8881 dan ~8920). Sanitasi diterapkan agar perhitungan return dan clustering bebas dari outlier artifisial.

```python
df = pd.read_csv('DATASET.csv')
df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values('Date').reset_index(drop=True)

# Sanitasi data typo desimal USDIDR (< 5000)
typo_count = (df['USDIDR'] < 5000).sum()
if typo_count > 0:
    print(f"Ditemukan {typo_count} baris typo desimal USDIDR (< 5000). Mengalikan 10 untuk normalisasi.")
    df.loc[df['USDIDR'] < 5000, 'USDIDR'] = df.loc[df['USDIDR'] < 5000, 'USDIDR'] * 10

# Filter data berlabel historis (2010-2024)
labeled_mask = df['Target_Volatilitas_Y'].notnull()
df_labeled = df[labeled_mask].copy()

# Tambahkan fitur kalkulasi dasar ke df_labeled
df_labeled['rate_spread'] = df_labeled['BI_rate'] - df_labeled['US_rate']
df_labeled['USDIDR_ret_20d'] = df_labeled['USDIDR'].pct_change(20).fillna(0)
df_labeled['SP500_ret_20d'] = df_labeled['SP500'].pct_change(20).fillna(0)
df_labeled['IHSG_ret_20d'] = df_labeled['IHSG'].pct_change(20).fillna(0)

print(f"Total baris data berlabel: {len(df_labeled):,} ({df_labeled['Date'].min().strftime('%Y-%m-%d')} s/d {df_labeled['Date'].max().strftime('%Y-%m-%d')})")
```

## 2. Pemilihan Variabel Indikator Rezim Makroekonomi
Untuk mendeteksi kondisi pasar secara holistik, kita memilih indikator kunci yang merepresentasikan:
1. **Sentimen Kepanikan Global:** `VIX`
2. **Siklus Kebijakan Moneter & Spread Likuiditas:** `rate_spread = BI_rate - US_rate` dan `US_rate`
3. **Tekanan Kurs Valuta Asing (FX):** `USDIDR_ret_20d` (return 20 hari kerja / 1 bulan)
4. **Momentum Ekuitas Global & Domestik:** `SP500_ret_20d` dan `IHSG_ret_20d`

```python
# Konstruksi matriks fitur rezim pasar
regime_cols = ['VIX', 'rate_spread', 'US_rate', 'BI_rate', 'USDIDR_ret_20d', 'SP500_ret_20d', 'IHSG_ret_20d']
regime_features = df_labeled[regime_cols].copy()

# Standarisasi fitur dengan StandardScaler
scaler = StandardScaler()
X_regime_scaled = scaler.fit_transform(regime_features)

print("Fitur indikator rezim berhasil dibentuk dengan dimensi:", X_regime_scaled.shape)
regime_features.describe().T[['mean', 'std', 'min', '50%', 'max']]
```

## 3. Penentuan Jumlah Klaster Optimal ($k$)
Kita mengevaluasi rentang $k \in [2, 7]$ menggunakan dua metrik kuantitatif:
- **Inertia (Elbow Method):** Mengukur kekompakan dalam klaster (*within-cluster sum of squares*).
- **Silhouette Score:** Mengukur pemisahan dan kejelasan batas antarklaster.

```python
k_range = range(2, 8)
inertias = []
sil_scores = []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=15)
    labels = km.fit_predict(X_regime_scaled)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X_regime_scaled, labels))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Elbow Method
axes[0].plot(k_range, inertias, marker='o', color='#1f77b4', linewidth=2.5, markersize=7)
axes[0].set_title('Elbow Curve (Inertia vs Jumlah Klaster k)', pad=12, fontweight='bold')
axes[0].set_xlabel('Jumlah Klaster (k)')
axes[0].set_ylabel('Inertia (Within-Cluster SSE)')
axes[0].set_xticks(k_range)

# Silhouette Score
axes[1].plot(k_range, sil_scores, marker='s', color='#2ca02c', linewidth=2.5, markersize=7)
axes[1].axvline(4, color='#d62728', linestyle='--', label='k=4 Optimal (Score Puncak: 0.2882)')
axes[1].set_title('Silhouette Score vs Jumlah Klaster k', pad=12, fontweight='bold')
axes[1].set_xlabel('Jumlah Klaster (k)')
axes[1].set_ylabel('Silhouette Score')
axes[1].set_xticks(k_range)
axes[1].legend(frameon=True)

plt.tight_layout()
plt.savefig('figures/04_optimal_k_elbow_silhouette.png', dpi=300, bbox_inches='tight')
plt.show()

for k, sil in zip(k_range, sil_scores):
    print(f"k = {k} -> Silhouette Score = {sil:.4f}")
```

## 4. Pembentukan Model Clustering ($k=4$) & Proyeksi 2D PCA
Nilai $k=4$ menghasilkan Silhouette Score tertinggi ($0.2882$) dengan interpretasi ekonomi yang sangat intuitif.
Untuk memvisualisasikan klaster dalam ruang berdimensi tinggi, kita menggunakan **Principal Component Analysis (PCA)** untuk mereduksi 7 dimensi fitur menjadi 2 komponen utama (PC1 dan PC2).

```python
# Pelatihan K-Means dengan k=4
optimal_k = 4
kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=20)
cluster_labels = kmeans.fit_predict(X_regime_scaled)
df_labeled['Regime_Cluster'] = cluster_labels

# Proyeksi PCA 2 Dimensi
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_regime_scaled)
df_labeled['PC1'] = X_pca[:, 0]
df_labeled['PC2'] = X_pca[:, 1]

print(f"Variance Explained PC1: {pca.explained_variance_ratio_[0]*100:.2f}%")
print(f"Variance Explained PC2: {pca.explained_variance_ratio_[1]*100:.2f}%")
print(f"Total Variance Explained (2 Komponen): {pca.explained_variance_ratio_.sum()*100:.2f}%")
```

```python
# Pemetaan Nama Rezim Berdasarkan Profil Ekonomi
regime_names = {
    0: 'Rezim 0: Global Panic & Crisis Shock',
    1: 'Rezim 1: Normal Moderate Transition',
    2: 'Rezim 2: High Spread & Low US Rate (QE Era)',
    3: 'Rezim 3: Aggressive Fed Tightening'
}
df_labeled['Regime_Name'] = df_labeled['Regime_Cluster'].map(regime_names)

palette = {
    'Rezim 0: Global Panic & Crisis Shock': '#e74c3c',          # Merah krisis
    'Rezim 1: Normal Moderate Transition': '#3498db',           # Biru normal
    'Rezim 2: High Spread & Low US Rate (QE Era)': '#f39c12',   # Oranye ekspansif
    'Rezim 3: Aggressive Fed Tightening': '#27ae60'             # Hijau stabil/ketat
}

plt.figure(figsize=(12, 7))
sns.scatterplot(
    data=df_labeled, x='PC1', y='PC2', hue='Regime_Name',
    palette=palette, alpha=0.75, s=40, edgecolor='none'
)

# Plot Centroids
centroids_pca = pca.transform(kmeans.cluster_centers_)
plt.scatter(
    centroids_pca[:, 0], centroids_pca[:, 1],
    c='black', marker='X', s=200, label='Cluster Centroids', edgecolor='white', linewidth=2
)

plt.title('Pemetaan 4 Rezim Pasar Makroekonomi dalam Ruang PCA 2 Dimensi', pad=14, fontweight='bold')
plt.xlabel(f"PC1 ({pca.explained_variance_ratio_[0]*100:.1f}% Variance Explained) — Suku Bunga & Spread Axis")
plt.ylabel(f"PC2 ({pca.explained_variance_ratio_[1]*100:.1f}% Variance Explained) — Global Fear (VIX) & Momentum Axis")
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', frameon=True)

plt.tight_layout()
plt.savefig('figures/05_pca_regime_clusters_2d.png', dpi=300, bbox_inches='tight')
plt.show()
```

## 5. Profil Makroekonomi & Perilaku Volatilitas IHSG per Rezim
Mari kita bedah karakteristik makro dan perilaku `Target_Volatilitas_Y` di setiap rezim pasar.

```python
# Kompilasi Tabel Rangkuman Karakteristik per Rezim
regime_summary = df_labeled.groupby('Regime_Name').agg(
    Jumlah_Hari=('Target_Volatilitas_Y', 'count'),
    Persentase_Data=('Target_Volatilitas_Y', lambda x: f"{len(x) / len(df_labeled) * 100:.1f}%"),
    Mean_Volatilitas_Y=('Target_Volatilitas_Y', lambda x: x.mean() * 100),
    Median_Volatilitas_Y=('Target_Volatilitas_Y', lambda x: x.median() * 100),
    Std_Volatilitas_Y=('Target_Volatilitas_Y', lambda x: x.std() * 100),
    Rata_VIX=('VIX', 'mean'),
    Rata_Spread=('rate_spread', 'mean'),
    Rata_US_Rate=('US_rate', 'mean'),
    Rata_BI_Rate=('BI_rate', 'mean'),
    Rata_Return_IHSG_20d=('IHSG_ret_20d', lambda x: x.mean() * 100)
).sort_values('Mean_Volatilitas_Y', ascending=False)

print("=== TABEL PROFIL REZIM PASAR & REALIZED VOLATILITY IHSG ===")
display(regime_summary.round(3))
```

## 6. Uji Signifikansi & Visualisasi Distribusi Volatilitas IHSG
Apakah perbedaan tingkat volatilitas antarrezim ini signifikan secara statistik? Kita menguji dengan uji non-parametrik **Kruskal-Wallis H-test**.

```python
# Uji Kruskal-Wallis antar kelompok rezim
groups = [group['Target_Volatilitas_Y'].values for _, group in df_labeled.groupby('Regime_Cluster')]
kw_stat, p_val = stats.kruskal(*groups)
print(f"Kruskal-Wallis H-statistic: {kw_stat:.2f}, p-value: {p_val:.4e}")
if p_val < 0.001:
    print("KESIMPULAN: Tingkat volatilitas IHSG berbeda secara signifikan (p < 0.001) antar rezim pasar!")

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Boxplot Distribusi Volatilitas
order_regimes = regime_summary.index.tolist()
sns.boxplot(
    data=df_labeled, y='Regime_Name', x='Target_Volatilitas_Y',
    order=order_regimes, palette=palette, ax=axes[0], showmeans=True,
    meanprops={"marker":"o", "markerfacecolor":"white", "markeredgecolor":"black", "markersize":"7"}
)
axes[0].set_title('Distribusi Realized Volatility IHSG 5-Hari per Rezim Pasar', pad=12, fontweight='bold')
axes[0].set_xlabel('Target_Volatilitas_Y (Titik putih = Mean)')
axes[0].set_ylabel('')

# KDE Plot (Densitas Probabilitas)
for reg_name in order_regimes:
    subset = df_labeled[df_labeled['Regime_Name'] == reg_name]['Target_Volatilitas_Y']
    sns.kdeplot(subset, ax=axes[1], label=reg_name.split(':')[1].strip(), color=palette[reg_name], linewidth=2.2)

axes[1].set_title('Kepadatan Distribusi Probabilitas Volatilitas IHSG (KDE)', pad=12, fontweight='bold')
axes[1].set_xlabel('Target_Volatilitas_Y')
axes[1].set_ylabel('Density')
axes[1].legend(loc='upper right', frameon=True)

plt.tight_layout()
plt.savefig('figures/06_volatilitas_ihsg_per_regime_boxplot.png', dpi=300, bbox_inches='tight')
plt.show()
```

## 7. Kronologi Rezim Pasar Historis (Timeline Shading 2010–2024)
Visualisasi memetakan perjalanan waktu 2010–2024, di mana latar belakang diwarnai sesuai rezim pasar yang aktif. Ini memperlihatkan secara jelas momen-momen krisis (*Global Panic*) seperti Krisis Utang Eropa 2011, Taper Tantrum 2013, Pandemi 2020, dan Siklus Pengetatan The Fed 2022–2024.

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(16, 8), sharex=True, gridspec_kw={'height_ratios': [2, 1.2]})

# Plot IHSG Level
ax1.plot(df_labeled['Date'], df_labeled['IHSG'], color='#2c3e50', linewidth=1.5, label='Indeks IHSG')
ax1.set_ylabel('IHSG Level (Poin)', fontweight='semibold')
ax1.set_title('Kronologi Historis Indeks IHSG & Target Volatilitas Berdasarkan Rezim Pasar (2010–2024)', pad=12, fontweight='bold')
ax1.legend(loc='upper left', frameon=True)

# Plot Target Volatilitas IHSG
ax2.plot(df_labeled['Date'], df_labeled['Target_Volatilitas_Y'] * 100, color='#8e44ad', linewidth=1.2, label='Target Realized Volatility (%)')
ax2.set_ylabel('Volatilitas (%)', fontweight='semibold')
ax2.set_xlabel('Tahun Observasi', fontweight='semibold')
ax2.legend(loc='upper left', frameon=True)

# Scatter titik warna rezim pada grafik
for reg_name, color in palette.items():
    mask = df_labeled['Regime_Name'] == reg_name
    ax1.scatter(df_labeled.loc[mask, 'Date'], df_labeled.loc[mask, 'IHSG'], color=color, s=2, alpha=0.4)
    ax2.scatter(df_labeled.loc[mask, 'Date'], df_labeled.loc[mask, 'Target_Volatilitas_Y'] * 100, color=color, s=2, alpha=0.4)

plt.tight_layout()
plt.savefig('figures/07_historical_market_regime_timeline.png', dpi=300, bbox_inches='tight')
plt.show()
```
